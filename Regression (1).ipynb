```python
import pandas as pd
data = pd.read_csv('/content/kc_house_data.csv')
data
df = data.copy()

print("="*50)
print("Initial dataset shape:", df.shape)
print("="*50)
```

```python
print("\nMissing values in each column:")
missing = df.isnull().sum()
print(missing)
df.dropna(inplace=True)
print(f"\nShape after dropping missing values: {df.shape}")
```

```python
print("\nData types before cleaning:")
print(df.dtypes)
df['date'] = pd.to_datetime(df['date'], format='mixed')
print("\nDate column converted to datetime")
print("\nFirst 5 rows after date conversion:")
print(df.head())
```

```python
df.dropna(subset=['date'], inplace=True)
print(f"\nShape after dropping rows with missing dates: {df.shape}")
```

```python
print("\nChecking for unrealistic values...")

unrealistic_bedrooms = df[(df['bedrooms'] < 0) | (df['bedrooms'] > 10)]
print(f"Rows with unrealistic bedrooms: {len(unrealistic_bedrooms)}")

df.loc[df['bedrooms'] < 0, 'bedrooms'] = 1
df.loc[df['bedrooms'] > 10, 'bedrooms'] = 5

unrealistic_bathrooms = df[(df['bathrooms'] < 0) | (df['bathrooms'] > 8)]
print(f"Rows with unrealistic bathrooms: {len(unrealistic_bathrooms)}")

df.loc[df['bathrooms'] < 0, 'bathrooms'] = 1
df.loc[df['bathrooms'] > 8, 'bathrooms'] = 3

unrealistic_price = df[df['price'] < 10000]
print(f"Rows with unrealistic price (< $10,000): {len(unrealistic_price)}")

df.loc[df['price'] < 10000, 'price'] = df['price'].median()

unrealistic_floors = df[(df['floors'] < 0) | (df['floors'] > 4)]
print(f"Rows with unrealistic floors: {len(unrealistic_floors)}")

df.loc[df['floors'] < 0, 'floors'] = 1
df.loc[df['floors'] > 4, 'floors'] = 2

print("\nUnrealistic values corrected")
```

```python
print("\n" + "="*50)
print("Checking for duplicates...")
print(f"Number of duplicate rows: {df.duplicated().sum()}")

duplicates = df[df.duplicated()]
if len(duplicates) > 0:
    print("\nFirst few duplicates:")
    print(duplicates.head())

df.drop_duplicates(inplace=True)
print(f"\nShape after removing duplicates: {df.shape}")
```

```python
print("\n" + "="*50)
print("Example: Filling missing values with mean...")

mean_sqft = df['sqft_living'].mean()
print(f"Mean of sqft_living: {mean_sqft:.2f}")

df['sqft_living'] = df['sqft_living'].fillna(mean_sqft)

print("Missing values in sqft_living filled with mean")
```

```python
df.reset_index(drop=True, inplace=True)

print("\n" + "="*50)
print("Final dataset info:")
print(f"Shape: {df.shape}")
print(f"Columns: {df.columns.tolist()}")
print("\nFirst 5 rows of cleaned data:")
print(df.head())
```

```python
print("\n" + "="*50)
print("SUMMARY OF DATA CLEANING:")
print("="*50)
print(f"✓ Initial rows: {len(data)}")
print(f"✓ Final rows: {len(df)}")
print(f"✓ Rows removed: {len(data) - len(df)}")
print(f"✓ Columns: {len(df.columns)}")
print(f"✓ No missing values: {df.isnull().sum().sum() == 0}")
print(f"✓ No duplicates: {df.duplicated().sum() == 0}")
print("="*50)
print("\nQuick statistics of cleaned data:")
print(df[['price', 'bedrooms', 'bathrooms', 'sqft_living']].describe())
```

```python
df.to_csv('/content/kc_house_data.csv', index=False)
print("\nCleaned data saved to '/content/kc_house_data.csv'")
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

df["date"] = pd.to_datetime(df["date"])
df["year"] = df["date"].dt.year
df["month"] = df["date"].dt.month
input = df.drop(["price", "date"], axis=1)
output = df["price"]


input_train , input_test , output_train ,output_test = train_test_split( input, output ,
                                                                         train_size=0.8 , random_state = 50)

linear=LinearRegression()
linear.fit(input_train,output_train)

prediction = linear.predict(input_test)

MAE=mean_absolute_error(output_test,prediction)
MSE=mean_squared_error(output_test,prediction)
RMSE=np.sqrt(MSE)
R2=r2_score(output_test,prediction)


```

```python
results = {"MAE": MAE , "MSE": MSE , "RMSE": RMSE , "R2": R2}
print(results)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.read_csv("/content/kc_house_data.csv")
plt.style.use('ggplot')
sns.set_style("whitegrid")
fig = plt.figure(figsize=(20, 16))

ax1 = fig.add_subplot(3, 3, 1)
ax1.hist(df["price"], bins=50, color='steelblue', edgecolor='black')
ax1.axvline(df["price"].mean(), color='red', linestyle='--', label=f'Mean: ${df["price"].mean():,.0f}')
ax1.axvline(df["price"].median(), color='green', linestyle='--', label=f'Median: ${df["price"].median():,.0f}')
ax1.set_xlabel("Price ($)"); ax1.set_ylabel("Number of Houses")
ax1.set_title("House Price Distribution"); ax1.legend()

ax2 = fig.add_subplot(3, 3, 2)
ax2.scatter(df["sqft_living"], df["price"], alpha=0.3, c='steelblue', s=10)
ax2.set_xlabel("Living Area (sqft)"); ax2.set_ylabel("Price ($)")
ax2.set_title("Living Area vs Price")

ax3 = fig.add_subplot(3, 3, 3)
df.boxplot(column='price', by='bedrooms', ax=ax3)
ax3.set_title("Price by Number of Bedrooms")
ax3.set_xlabel("Bedrooms"); ax3.set_ylabel("Price ($)")

ax4 = fig.add_subplot(3, 3, 4)
important_cols = ['price', 'sqft_living', 'grade', 'sqft_above', 'sqft_living15',
                  'bathrooms', 'view', 'sqft_basement', 'bedrooms', 'yr_built']
corr_matrix = df[important_cols].corr()
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', center=0, fmt='.2f',
            square=True, ax=ax4, cbar_kws={'shrink': 0.8})
ax4.set_title("Correlation Matrix")

ax5 = fig.add_subplot(3, 3, 5)
corr_with_price = df[important_cols].corr()["price"].sort_values(ascending=False)
top_features = corr_with_price.drop('price')
colors = ['green' if x > 0 else 'red' for x in top_features.values]
ax5.barh(top_features.index, top_features.values, color=colors)
ax5.set_xlabel("Correlation Coefficient")
ax5.set_title("Feature Correlation with Price")
ax5.axvline(0, color='black', linewidth=0.5)

ax6 = fig.add_subplot(3, 3, 6)
df.boxplot(column='price', by='grade', ax=ax6)
ax6.set_title("Price by Construction Grade")
ax6.set_xlabel("Grade (1-13)"); ax6.set_ylabel("Price ($)")

ax7 = fig.add_subplot(3, 3, 7)
ax7.scatter(df["yr_built"], df["price"], alpha=0.2, c='darkblue', s=8)
ax7.set_xlabel("Year Built"); ax7.set_ylabel("Price ($)")
ax7.set_title("Year Built vs Price")

ax8 = fig.add_subplot(3, 3, 8)
waterfront_prices = [df[df['waterfront']==0]['price'], df[df['waterfront']==1]['price']]
ax8.boxplot(waterfront_prices, labels=['No Waterfront', 'Has Waterfront'])
ax8.set_ylabel("Price ($)"); ax8.set_title("Waterfront Impact on Price")

ax9 = fig.add_subplot(3, 3, 9)
df['bedrooms'].value_counts().sort_index().plot(kind='bar', ax=ax9, color='coral', edgecolor='black')
ax9.set_xlabel("Number of Bedrooms"); ax9.set_ylabel("Number of Houses")
ax9.set_title("Bedrooms Distribution")

plt.tight_layout()
plt.show()
```
